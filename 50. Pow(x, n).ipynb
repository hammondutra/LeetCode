```python
class Solution:
    def myPow(self, x: float, n: int) -> float:
        # Exponent Conditions
        if n == 0:
            return 1
        if n == 1:
            return x
        if n < 0:
            return 1 / self.myPow(x, -n)
        
        # Base Conditions
        if x == 0:
            return 0
        if n % 2 == 0:
            half = self.myPow(x, n//2)
            return half * half
        else: # n % 2 == 1:
            half = self.myPow(x, (n-1)//2)
            return x * half * half
```
