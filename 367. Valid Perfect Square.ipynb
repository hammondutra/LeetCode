```python
class Solution:
    def isPerfectSquare(self, num: int) -> bool:
        if num == 1:
            return True
        
        l = 1
        u = num//2

        while l <= u:
            g = (l + u)//2
            g_sq = g**2

            if g_sq == num:
                return True
            elif g_sq < num:
                l = g + 1
            elif g_sq > num:
                u = g - 1
        
        return False
```
