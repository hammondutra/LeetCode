```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def hasCycle(self, head: Optional[ListNode]) -> bool:
        if not head:
            return False
        
        seen = set()
        curr = head

        while curr: # while there is a next pointer
            if curr not in seen:
                seen.add(curr)
            else: # head.val in seen
                return True
            curr = curr.next
        
        return False
```
