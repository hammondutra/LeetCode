```python
class Solution:
    def plusOne(self, digits: List[int]) -> List[int]:
        if not digits:
            return []

        if digits[-1] != 9:
            digits[-1] += 1
            return digits
        else:
            digits[-1] = 0
        
        for i in range(2, len(digits)+1):
                if digits[-i] == 9:
                    digits[-i] = 0
                else: # digits[-i] != 9:
                    digits[-i] += 1
                    return digits

        digits.insert(0, 1)
        return digits
```
