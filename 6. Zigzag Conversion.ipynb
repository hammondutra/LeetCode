```python
class Solution:
    def convert(self, s: str, numRows: int) -> str:
        # Special Cases
        if numRows == 1 or numRows >= len(s):
            return s
        
        if numRows == 2:
            first = ""
            second = ""
            for i in range(len(s)):
                if i % 2 == 0:
                    first += s[i]
                else: # i % 2 == 1
                    second += s[i]
            return first + second
        
        # For numRows > 2:
        lines = [] # lst[str] where lines[i] contains the ith line
        for i in range(numRows):
            lines.append("")

        for i in range(len(s)): # i = 3, numRows = 4
            result = i % (numRows*2 - 2) # result = 3 % (4*2 - 2) = 3 % 6 == 3
            if result <= numRows-1: # Ida
                lines[result] += s[i]
            else: # Volta
                lines[(numRows * 2 - 2) - result] += s[i]
        
        return "".join(lines)
```
