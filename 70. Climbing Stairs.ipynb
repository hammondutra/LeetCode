```python
class Solution:
    calculated = {}

    def climbStairs(self, n: int) -> int:
        if n == 1:
            return 1
        if n == 2:
            return 2
        elif n in self.calculated:
            return self.calculated[n]
        
        else: # Recursive step
            self.calculated[n] = self.climbStairs(n-1) + self.climbStairs(n-2)
            return self.calculated[n]
```
