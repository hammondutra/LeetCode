```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        carry = 0
        head, curr1, curr2 = l1, l1, l2
        prev = None
        
        while curr1 or curr2:
            if not curr1:
                # Create new node attached to end of list
                prev.next = ListNode()
                curr1 = prev.next
                sum = curr2.val + carry
            elif not curr2:
                sum = curr1.val + carry
            else:
                sum = curr1.val + curr2.val + carry
            
            if sum > 9:
                carry = 1
            else:
                carry = 0
            
            # Take the last digit and reassign curr1.val
            curr1.val = sum % 10
            prev = curr1

            # Only advance them if their not already None
            if curr1:
                curr1 = curr1.next
            if curr2:
                curr2 = curr2.next
        
        # After both loops end
        if carry:
            prev.next = ListNode(1)
        
        return head
```
