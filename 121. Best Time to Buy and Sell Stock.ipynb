```python
class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        if len(prices) == 0:
            return 0
        
        max_profit = 0
        min_price = float("inf")

        for i in range(len(prices)):
            if prices[i] < min_price:
                min_price = prices[i]

            elif prices[i] - min_price > max_profit: # Check what the profit would be if we compared selling today (at a higher price) vs. our best value so far
                max_profit = prices[i] - min_price
                
        return max_profit
```
