```python
class Solution:
    def merge(self, nums1: List[int], m: int, nums2: List[int], n: int) -> None:
        i = m - 1
        j = n - 1
        insertion_pointer = m + n - 1

        while j >= 0:
            if i >= 0 and nums1[i] > nums2[j]:
                nums1[insertion_pointer] = nums1[i]
                i -= 1
            else:
                nums1[insertion_pointer] = nums2[j]
                j -= 1
            
            insertion_pointer -= 1
```
