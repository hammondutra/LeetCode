```python
class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        def reverser(left=0, right=len(nums)-1):
            while left < right:
                nums[left], nums[right] = nums[right], nums[left]
                left += 1
                right -= 1
        
        reverser(0, len(nums)-1) # Reverses the full array
        reverser(0, k % len(nums) - 1)
        reverser(k % len(nums), len(nums) - 1)
```
