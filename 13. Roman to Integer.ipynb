```python
class Solution:
    def romanToInt(self, s: str) -> int:
        conversion_dict = {"I": 1, "V": 5, "X": 10, "L": 50, "C": 100, "D": 500, "M": 1000}
        total_sum = 0

        for i in range(len(s)-1):
            if conversion_dict[s[i]] < conversion_dict[s[i+1]]:
                total_sum -= conversion_dict[s[i]]
                    
            else: # conversion_dict[s[i]] > conversion_dict[s[i+1]] or they are the same number
                total_sum += conversion_dict[s[i]]
        
        total_sum += conversion_dict[s[-1]] # The last character always gets added regardless

        return total_sum
```
