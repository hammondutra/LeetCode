```python
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        i = 2 # read pointer
        k = 2 # write pointer

        while i < len(nums): # Start iterating from 2nd element in array
            if nums[i] != nums[k-2]:
                nums[k] = nums[i]
                k += 1
                i += 1

            else: # nums[i] == nums[k-2]
                i += 1
        
        return k
```
