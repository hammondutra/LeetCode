```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        length_s = len(s)
        length_t = len(t)

        if (not s) or (length_s != length_t):
            return False
        
        if not t:
            return True

        density_dict = {}
        for i in range(length_s):
            if s[i] not in density_dict:
                density_dict[s[i]] = 1

            else: # s[i] in density_dict:
                density_dict[s[i]] += 1
        
        for i in range(length_t):
            if t[i] not in density_dict:
                return False
            
            else: # t[i] in density_dict
                density_dict[t[i]] -= 1
                if density_dict[t[i]] < 0:
                    return False

        # If nothing of this happens while iterating through t, it means that both words are anagrams of each other
        return True 
```
