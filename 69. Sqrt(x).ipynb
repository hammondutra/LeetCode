```python
class Solution:
    def mySqrt(self, x: int) -> int:
        if x == 0:
            return 0

        left = 1
        right = x
        closest = 1

        while left <= right:
            mid = (left + right) // 2
            
            if mid * mid == x:
                return mid
            
            elif mid * mid < x:
                closest = mid
                left = mid + 1
            
            else: # mid * mid > x:
                right = mid - 1
        
        # If we exit the loop and no exact match has been found, we return the closest from below
        return closest
```
