```python
class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        if len(nums) == 0 or not nums:
            return 0

        lower = 0
        upper = len(nums)

        while lower < upper:
            guess = (lower + upper) // 2
                
            if nums[guess] == target:
                return guess
            elif nums[guess] < target:
                lower = guess + 1
            else:
                upper = guess
            
        return lower
```
