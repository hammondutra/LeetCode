```python
class Solution:
    def canConstruct(self, ransomNote: str, magazine: str) -> bool:
        if not magazine:
            return False
        
        if not ransomNote:
            return True

        density_dict = {}

        for i in range(len(magazine)): # Time complexity O(n)
            if magazine[i] not in density_dict:
                density_dict[magazine[i]] = 1
            
            else:
                density_dict[magazine[i]] += 1
        
        # Up until here, we've constructed the density dictionary
        for i in range(len(ransomNote)): # Time complexity O(m)
            if (ransomNote[i] not in density_dict) or (density_dict[ransomNote[i]] <= 0):
                return False
            elif density_dict[ransomNote[i]] > 0:
                density_dict[ransomNote[i]] -= 1
        
        return True
```
