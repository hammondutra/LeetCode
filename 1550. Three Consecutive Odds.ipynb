```python
class Solution:
    def threeConsecutiveOdds(self, arr: List[int]) -> bool:
        odd_counter = 0
        for i in arr:
            if i % 2 == 1:
                odd_counter += 1
                if odd_counter == 3:
                    return True
                    break
            else:
                odd_counter = 0
        return False
```
