```python
class Solution:
    def isHappy(self, n: int) -> bool:
        if sum([int(str(n)[i])**2 for i in range(len(str(n)))]) == 1:
            return True
        
        my_dict = {}
        number = n
        squared_digits = [int(str(n)[i])**2 for i in range(len(str(n)))]

        while sum(squared_digits) != 1:
            if sum(squared_digits) not in my_dict:
                my_dict[sum(squared_digits)] = 1
                squared_digits = [int(str(sum(squared_digits))[i])**2 for i in range(len(str(sum(squared_digits))))]    
                print(sum(squared_digits))
            
            else: # If this value is already in my_dict -> cycle was detected
                return False
        
        return True # If we exited loop, sum of the squared digits equals 1 -> Happy number
```
