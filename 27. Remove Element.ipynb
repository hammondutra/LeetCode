```python
class Solution:
    def removeElement(self, nums: List[int], val: int) -> int:
        if not nums:
            return 0
        
        count = 0

        for i in range(len(nums) -1, -1, -1):
            if nums[i] == val:
                count += 1
                nums.append(val)
                nums.pop(i)

        return len(nums) - count
```
