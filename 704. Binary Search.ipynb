```python
class Solution:
    def search(self, nums: List[int], target: int) -> int:
        low = 0
        high = len(nums) - 1

        while low <= high:
            guess = (low + high) // 2

            if nums[guess] == target:
                return guess # return index of target
            
            elif nums[guess] < target: 
                low = guess + 1
            
            elif nums[guess] > target:
                high = guess - 1
        
        return -1
```
