```python
class Solution:
    def isSubsequence(self, s: str, t: str) -> bool:
        if (len(s) == 0) or (s == t):
            return True
        
        if not t:
            return False

        curr = 0
        for i in range(len(t)):
            if s[curr] == t[i]:
                curr += 1
                if curr == len(s):
                    return True
        
        # If loop is exited, some matches were not found
        return False
```
