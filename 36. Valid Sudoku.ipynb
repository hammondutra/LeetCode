```python
class Solution:
    def isValidSudoku(self, board: List[List[str]]) -> bool:
        boxes = {0: set(), 1: set(), 2: set(), 3: set(), 4: set(), 5: set(), 6: set(), 7: set(), 8: set()}
        columns = {0: set(), 1: set(), 2: set(), 3: set(), 4: set(), 5: set(), 6: set(), 7: set(), 8: set()}

        for y in range(9):
            rows = set()
            for x in range(9):
                if board[y][x] != ".":
                    # Tracking the box of element (x, y)
                    x_box = x // 3
                    y_box = y // 3

                    # Checking conditions 1 through 3:
                    if (board[y][x] in rows) or (board[y][x] in columns[x]) or (board[y][x] in boxes[x_box + 3*y_box]):
                        return False
                    
                    else: # Passed all conditions
                        rows.add(board[y][x])
                        columns[x].add(board[y][x])
                        boxes[x_box + 3*y_box].add(board[y][x])

        return True
```
