```python
class Solution:
    def maxArea(self, height: List[int]) -> int:
        lower = 0
        upper = len(height) - 1
        best = 0

        while lower < upper:
            current = (upper - lower) * min(height[lower], height[upper])

            best = max(best, current)

            if height[lower] < height[upper]:
                    lower += 1
            else: # height[lower] >= height[upper]
                upper -= 1
        
        return best
```
