```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        comp_dict = {}

        for i in range(len(nums)):
            comp_dict[nums[i]] = i
        
        for i in range(len(nums)):
            complement = target - nums[i] # Determine the target for each i
            if (complement in comp_dict) and (comp_dict[complement]) != i:
                return [i, comp_dict[complement]] # Return index of element + index of complement
```
