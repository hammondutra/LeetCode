```python
class Solution:
    def isIsomorphic(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False
        
        map_dict = {}
        i = 0

        while i <= len(t) - 1:
            if (s[i] not in map_dict):

                if t[i] not in map_dict.values():
                    map_dict[s[i]] = t[i]
                
                else:
                    return False
            
            elif (s[i] in map_dict) and (map_dict[s[i]] != t[i]):
                return False

            i += 1
        
        return True
```
