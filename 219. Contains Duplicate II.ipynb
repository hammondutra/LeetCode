```python
class Solution:
    def containsNearbyDuplicate(self, nums: List[int], k: int) -> bool:
        my_dict = {}

        for i in range(len(nums)):
            if(nums[i] in my_dict) and (i - my_dict[nums[i]] <= k): # This already checks both conditions
                return True
            
            my_dict[nums[i]] = i
        
        return False
```
