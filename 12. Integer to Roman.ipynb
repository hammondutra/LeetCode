```python
class Solution:
    def intToRoman(self, num: int) -> str:
        if not num:
            return None
        
        values_dict = {1: "I", 5: "V", 10: "X", 50: "L", 100: "C", 500: "D", 1000: "M"}
        result = []

        # Go from character to character, from left to right:
        str_num = str(num)
        no_of_digits = len(str_num)

        for i in range(no_of_digits):
            char = int(str_num[i])
            place_val = 10**(no_of_digits - i - 1)
            if char <= 5:
                if char in [1, 2, 3]:
                    for _ in range(char):
                        result.append(values_dict[place_val])

                elif char == 4:
                    result.append(values_dict[place_val] + values_dict[5 * place_val])
                
                elif char == 5:
                    result.append(values_dict[5 * place_val])

            elif char > 5:
                if char in [6, 7, 8]:
                    result.append(values_dict[5 * place_val])
                    for _ in range(char-5):
                        result.append(values_dict[place_val])

                elif char == 9:
                    result.append(values_dict[place_val] + values_dict[10*place_val])
        
        return "".join(result)
```
