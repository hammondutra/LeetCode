```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        if not x:
            return True
        
        x = str(x)
        s, e = 0, len(x) - 1

        while s < e:
            if x[s] != x[e]: # If there is a discrepancy
                return False
            s += 1
            e -= 1
        return True
```
