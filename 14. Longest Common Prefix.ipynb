```python
class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:
        if len(strs) == 0:
            return ""

        if len(strs) == 1:
            return strs[0]

        shared = []
        len_min_word = min(len(strs[i]) for i in range(len(strs)))

        for i in range(len_min_word):
            for j in range(len(strs)):
                if strs[1][i] != strs[j][i]:
                    return "".join(shared)
                
            shared.append(strs[1][i])
        
        return "".join(shared)
```
