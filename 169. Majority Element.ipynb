```python
class Solution:
    def majorityElement(self, nums: List[int]) -> int:
        if len(nums) == 0:
            return None
        
        if len(nums) == 1:
            return nums[0]


        frequency_dict = {}
        goal = (len(nums) // 2) + 1

        for i in range(len(nums)):
            if nums[i] in frequency_dict: # Means if we've already seen this number
                frequency_dict[nums[i]] += 1

                if frequency_dict[nums[i]] >= goal:
                    return nums[i]

            else: # Means if we've not yet seen this number
                frequency_dict[nums[i]] = 1

                if frequency_dict[nums[i]] >= goal:
                    return nums[i]
        
        return nums[i]
```
