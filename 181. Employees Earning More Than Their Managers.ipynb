```python
# Write your MySQL query statement below

# Determines "Employee" to be the only column in the output table
SELECT
    employee.name as Employee

# Creates two aliases for the tables (self-join)
FROM Employee as employee
JOIN Employee as manager

    # Then, compares the information of the manager with that of the employee
    ON employee.managerId = manager.id

# And filters just for those where the salary of the employee is greater than that of the manager
WHERE employee.salary > manager.salary
```
