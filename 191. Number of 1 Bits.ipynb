```python
class Solution:
    def hammingWeight(self, n: int) -> int:
        max_power = 0
        count = 0
        number = n

        while 2**max_power < n:
            max_power += 1

        while number > 0:
            if number - 2**max_power >= 0:
                count += 1
                number -= 2**max_power
            max_power -= 1
        
        return count
```
