```python
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        global_best = float("-inf")
        curr_sum = 0

        for num in nums:
            curr_sum += num
        
            if curr_sum > global_best:
                global_best = curr_sum

            if curr_sum < 0:
                curr_sum = 0
        
        return global_best
```
