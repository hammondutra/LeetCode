```python
class Solution:
    def wordPattern(self, pattern: str, s: str) -> bool:
        if not (pattern and s):
            return True

        word_count = 1
        char = 0
        while char < len(s):
            if s[char] == ' ':
                word_count += 1
            char += 1

        if (not pattern) or (not s) or (word_count != len(pattern)):
            return False
        
        seen_words = set()
        match_dict = {}
        start = -1
        end = -1

        for i in range(len(pattern)):
            start = end + 1
            end = start

            while end < len(s) and s[end] != ' ':
                end += 1 
            
            curr_word = s[start:end]

            if pattern[i] not in match_dict:
                if curr_word in seen_words:
                    return False
                
                match_dict[pattern[i]] = s[start:end]
                seen_words.add(curr_word)
            
            elif match_dict[pattern[i]] != s[start:end]:
                return False
        
        # If nothing happens, the mapping is 1-1 and onto
        return True
```
