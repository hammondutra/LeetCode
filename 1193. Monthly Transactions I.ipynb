```python
# Write your MySQL query statement below
SELECT
    DATE_FORMAT(trans_date, '%Y-%m') AS month, -- sorts by month
    country, -- sorts by country
    COUNT(*) as trans_count, -- counts all transactions

    SUM(CASE WHEN state = 'approved' THEN 1 ELSE 0 END) as approved_count, -- counts all approved transactions

    SUM(amount) AS trans_total_amount, -- counts total value of transactions

    SUM(CASE WHEN state = 'approved' THEN amount ELSE 0 END) as approved_total_amount -- counts total value of approved transactions

FROM Transactions
GROUP BY
    DATE_FORMAT(trans_date, '%Y-%m'), -- Groups by month
    country; -- Groups by country
```
