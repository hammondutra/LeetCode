```python
class Solution:
    def summaryRanges(self, nums: List[int]) -> List[str]:
        if not nums:
            return []
        
        my_lst = []
        left = nums[0]

        for i in range(1, len(nums) + 1): # Do len(nums) iterations
            if (i == len(nums)) or (nums[i] - nums[i-1] != 1):
                right = nums[i-1]
            
                if left == right:
                    my_lst.append(f"{left}")
                
                else: # left != right
                    my_lst.append(f"{left}->{right}")
                
                if i < len(nums):
                    left = nums[i] # Chooses the next element to be the beginning of the new interval

        return my_lst
```
