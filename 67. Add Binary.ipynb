```python
class Solution:
    def addBinary(self, a: str, b: str) -> str:
        result = []
        i = 1
        carry = 0

        while i <= max(len(a), len(b)):
            val1 = 0 if i > len(a) else int(a[-i])
            val2 = 0 if i > len(b) else int(b[-i])
            
            if (val1 + val2 + carry) == 0:
                result.append("0")
                carry = 0
            elif (val1 + val2 + carry) == 1:
                result.append("1")
                carry = 0
            else: # (val1 + val2 + carry) > 1: # Sum will result in a carry
                result.append(str((val1 + val2 + carry) % 2))
                carry = 1
            
            i += 1
        
        if carry: # If there is a carry in the end
            result.append("1")
        
        result.reverse()
        return "".join(result)
```
